<h3 style="color:#0000FF">Lineare Algebra mit Python</h3>

<div style="width: 80%;">

Zweidimensionale Arrays sind die Matrizen der linearen Algebra und NumPy beherscht natürlich auch die Lineare Algebra.

Das Produkt zweier Matrizen wird mit dem Operationszeichen **@** berechnet.

```python
import numpy as np
### Lineare Algebra
A = np.array([[1,2,3],[4,5,6]])
B = np.array([[1,2],[3,4],[5,6]])
print(A @ B)
```
